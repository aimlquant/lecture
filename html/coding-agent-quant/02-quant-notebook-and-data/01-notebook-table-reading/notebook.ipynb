# 처음 만드는 주피터 노트북

노트북은 설명을 읽는 **설명 셀**과 파이썬 명령을 실행하는 **코드 셀**로 이루어져 있습니다.

시작하기 전에 화면 위쪽의 커널 선택 메뉴에서 **Python 3**를 선택하세요. 커널은 코드를 실행하고 변수의 값을 기억하는 실행 엔진입니다. 여러 환경이 보이면 수업에서 사용하는 Python 3 환경을 선택하세요.

코드 셀을 선택한 뒤 **셀 왼쪽의 실행 버튼(▶)**을 누르면 그 셀을 실행할 수 있습니다. 실행 버튼이 보이지 않으면 셀 위에 마우스를 올려 보세요. `Shift + Enter`로 실행하고 다음 셀로 이동할 수도 있습니다.

셀은 **위에서 아래로 차례대로** 실행하세요. 뒤쪽 셀은 앞쪽 셀에서 정한 값을 사용하기 때문입니다. 커널을 다시 시작하면 기억하던 값이 사라지므로 첫 코드 셀부터 다시 실행하세요. 이 파일은 검증 결과를 저장한 비교 답안입니다. 내 노트북을 먼저 실행한 뒤 비교하세요.


## 1. 첫 문장 출력하기

`print()`는 괄호 안의 내용을 화면에 보여 줍니다. 아래 코드 셀 왼쪽의 실행 버튼을 누르면 셀 아래에 성공 문장이 나타납니다.


```python
print("첫 주피터 노트북 실행에 성공했습니다.")
```

## 2. 두 가격을 변수에 저장하기

변수는 값을 이름표에 담아 기억하는 방법입니다. `=`는 오른쪽 값을 왼쪽 이름에 저장한다는 뜻입니다.

기준가격은 100, 비교가격은 105로 정합니다. 이 값은 **학습용 가상 가격**이며 실제 시장 가격이 아닙니다. 두 값을 저장하고 각각 출력해 봅니다.


```python
기준가격 = 100
비교가격 = 105
print("기준가격:", 기준가격)
print("비교가격:", 비교가격)
```

## 3. 가격 변화율 계산하기

`(비교가격 / 기준가격 - 1) * 100`으로 기준가격에 비해 가격이 몇 퍼센트 변했는지 계산하고, 결과를 `변화율`에 저장합니다.

먼저 2단계 셀을 실행해야 두 가격을 사용할 수 있습니다. 출력 코드의 `:.1f`는 소수점 아래 한 자리까지 표시하라는 뜻입니다. 실행하면 `가격 변화: 5.0%` 형식으로 나타납니다.


```python
변화율 = (비교가격 / 기준가격 - 1) * 100
print(f"가격 변화: {변화율:.1f}%")
```

## 4. 기존 CSV 파일을 pandas로 읽기

`pandas`는 표 형태의 데이터를 다루는 도구입니다. `pd.read_csv()`로 기존 CSV를 읽어 `가격표`에 담습니다. 선택한 Python 3 환경에 pandas가 준비되어 있어야 합니다.

경로 `../data/kodex200-price-fixture.csv`의 `..`는 한 단계 위 폴더를 뜻합니다. 이 노트북이 있는 `notebooks` 폴더를 기준으로, 상위 폴더의 `data` 안에 있는 파일을 읽습니다.

`shape`는 표 전체의 행 수와 열 수를 알려 줍니다. `[["Date", "Close"]]`로 날짜와 종가 열을 고르고, `head(4)`로 앞 네 행을 보여 줍니다. 마지막 줄의 표는 노트북에서 자동으로 표시됩니다. CSV는 읽기만 하며 변경하지 않습니다.


```python
import pandas as pd

가격표 = pd.read_csv("../data/kodex200-price-fixture.csv")
print(f"전체 행 수: {가격표.shape[0]}, 전체 열 수: {가격표.shape[1]}")
가격표[["Date", "Close"]].head(4)
```
